```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Inspecting the data
```

```python
df = pd.read_csv('uber.csv')
```

```python
print(df.head())
```

```python
#Removing an unnecessary column unnamed : 0
df = df.drop(columns=["Unnamed: 0"])
```

```python
df = df.drop_duplicates()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
#Converting datetime from str to datetime
df['pickup_datetime'] = pd.to_datetime(df['pickup_datetime'],utc=True)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
#Outliers
sns.boxplot(data=df['fare_amount'])
plt.title("Outliers in fare_amount")
plt.show()
```

```python
# Inter-quantile range Method
Q1 = df['fare_amount'].quantile(0.25)
Q3 = df['fare_amount'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

outliers = df[(df['fare_amount'] < lower) | (df['fare_amount'] > upper)]
df = df[(df['fare_amount'] >= lower) & (df['fare_amount'] <= upper)]
```

```python
sns.boxplot(data=df['fare_amount'])
plt.title("Outliers in fare_amount")
plt.show()
```

```python
num_cols = df.select_dtypes(include=['int64','float64']).columns
```

```python
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR
    
    outliers = df[(df[col] < lower) | (df[col] > upper)] 
    sns.boxplot(data=df[col])
    plt.title(f"Outliers in {col}")
    plt.show()
```

```python
#Correlation
corr = df.corr(numeric_only=True)
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.show()
```

```python
#Extracting features
df['pickup_datetime'] = pd.to_datetime(df['pickup_datetime'],utc=True)
df['hour'] = df['pickup_datetime'].dt.hour
df['month'] = df['pickup_datetime'].dt.month
df['day'] = df['pickup_datetime'].dt.day
df['day_of_week'] = df['pickup_datetime'].dt.dayofweek

df.head()
```

```python
X = df.drop(columns=['key','fare_amount','pickup_datetime'])
Y = df['fare_amount']
```

```python
print(X.head())
print(Y.head())
```

```python
# Splitting the dataset
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
#Linear Regression

from sklearn.linear_model import LinearRegression
model1 = LinearRegression()
```

```python
model1.fit(x_train,y_train)
```

```python
y_pred = model1.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
print(f"Mean Absolute Error : {mean_absolute_error(y_test,y_pred)}")
```

```python
print(f"Mean Squared Error : {mean_squared_error(y_test,y_pred)}")
```

```python
print(f"R2 Score : {r2_score(y_test,y_pred)}")
```

```python
plt.scatter(y_test, y_pred)
plt.plot(y_test,y_test)

plt.xlabel("Actual Fare")
plt.ylabel("Predicted Fare")
plt.title("Linear Regression: Actual vs Predicted Fare")

plt.show()
```

```python
# Random Forest Regression

from sklearn.ensemble import RandomForestRegressor
```

```python
model2 = RandomForestRegressor(
    n_estimators=100,
    max_depth=15,
    random_state=42
)
```

```python
model2.fit(x_train,y_train)
```

```python
y_pred2 = model2.predict(x_test)
```

```python
print(f"Mean Absolute Error : {mean_absolute_error(y_test,y_pred2)}")
```

```python
print(f"Mean Squared Error : {mean_squared_error(y_test,y_pred2)}")
```

```python
print(f"R2 Score : {r2_score(y_test,y_pred2)}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.scatter(y_test, y_pred2, alpha=0.3)

plt.xlabel("Actual Fare")
plt.ylabel("Predicted Fare")
plt.title("Random Forest: Actual vs Predicted Fare")

plt.show()
```
