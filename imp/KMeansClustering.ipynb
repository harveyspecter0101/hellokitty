```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('sales_data_sample.csv', encoding='latin1')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
# Extracting the features
X = df[['QUANTITYORDERED','PRICEEACH','SALES','MSRP']]
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
X.isnull().sum()
```

```python
X = X.dropna()
```

```python
# Scaling the data
s = StandardScaler()
X = s.fit_transform(X)
```

```python
# Elbow Point
wcss = []
for i in range(1,11):
    km = KMeans(n_clusters=i,random_state=42,n_init=10)
    km.fit(X)
    wcss.append(km.inertia_)

plt.plot(range(1,11),wcss,marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.title('Elbow Method')
plt.show()
```

```python
km = KMeans(n_clusters=4,random_state=42,n_init=10)
df['Cluster'] = km.fit_predict(X)
df[['QUANTITYORDERED','PRICEEACH','SALES','MSRP','Cluster']].head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.scatterplot(data=df, x='QUANTITYORDERED', y='SALES',
                hue='Cluster')

plt.title('K-Means Clustering')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.scatterplot(data=df, x='PRICEEACH', y='SALES',
                hue='Cluster')

plt.title('K-Means Clustering')
plt.show()
```

```python
from sklearn.metrics import silhouette_score

print("Silhouette Score:", silhouette_score(X, df['Cluster']))
```

```python
# Hierarchical Clustering
from sklearn.cluster import AgglomerativeClustering
```

```python
hc = AgglomerativeClustering(n_clusters=4,linkage='ward')
```

```python
df['HC'] = hc.fit_predict(X)
```

```python
df[['QUANTITYORDERED','PRICEEACH','SALES','MSRP','HC']].head()
```

```python
import seaborn as sns

sns.scatterplot(
    data=df,
    x='QUANTITYORDERED',
    y='SALES',
    hue='HC'
)

plt.title('Hierarchical Clustering')
plt.show()
```

```python
from sklearn.metrics import silhouette_score
```

```python
print(f"Silhoutte Score : {silhouette_score(X,df['HC'])}")
```
